```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Name": ["Amina", "Daniyar", "Alina", "Maxim", "Sara"],
    "Age": [19, 21, 22, 20, 23],
    "Score": [78, 85, 90, 65, 88]
}

df = pd.DataFrame(data)
```

```python
# task 6

plt.plot(df["Name"], df["Score"], marker="o")
plt.xlabel("Students")
plt.ylabel("Score")
plt.title("Student Scores (Line Plot)")
plt.grid(True)
plt.show()
```

```python
# task 7

plt.bar(df["Name"], df["Score"])
plt.xlabel("Students")
plt.ylabel("Score")
plt.title("Student Scores (Bar Chart)")
plt.show()
```
