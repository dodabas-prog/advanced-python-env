```python
# task 3

data = {
    "Name": ["Amina", "Daniyar", "Alina", "Maxim", "Sara"],
    "Age": [19, 21, 22, 20, 23],
    "Score": [78, 85, 90, 65, 88]
}

df = pd.DataFrame(data)
print(df)
```

```python
# task 4

print("First 3 rows:")
print(df.head(3))

print("\nSummary statistics:")
print(df.describe())

print("\nAverage score:")
print(df["Score"].mean())
```

```python
# task 5

print("\nStudents older than 20:")
print(df[df["Age"] > 20])

print("\nStudents with score above 80:")
print(df[df["Score"] > 80])
```
